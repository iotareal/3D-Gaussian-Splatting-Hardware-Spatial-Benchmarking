```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
logs = Path(Path.cwd().parent) / "assets" / "GPU-logs-csv"

flower_rtx3060_4 = pd.read_csv(logs/"flower_rtx3060_12G_4.csv")
flower_rtx3060_8 = pd.read_csv(logs/"flower_rtx3060_12G_8.csv")
r_flower_rtx3060_4 = pd.read_csv(logs/"results_flower_rtx3060_12G_4.csv")
r_flower_rtx3060_8 = pd.read_csv(logs/"results_flower_rtx3060_12G_8.csv")
lego_rtx3050_8 = pd.read_csv(logs/"lego_rtx3050_4G_8.csv")
r_lego_rtx3060_8 = pd.read_csv(logs/"results_lego_rtx3050_4G_8.csv")
tank_rtx3060_4 = pd.read_csv(logs/"tank_rtx3060_12G_4.csv")
tank_rtx3060_8 = pd.read_csv(logs/"tank_rtx3060_12G_4.csv")
r_tank_rtx3060_4 = pd.read_csv(logs/"results_tank_rtx3060_12G_4.csv")
r_tank_rtx3060_8 = pd.read_csv(logs/"results_tank_rtx3060_12G_8.csv")

# Create a figure with 3 separate subplots vertically
fig, axes = plt.subplots(3, 1, figsize=(10, 12), sharex=True)
fig.suptitle(f'Hardware Performance Profile: {"lego_rtx3050_4G_8.csv"}', fontsize=16, y=0.98)

# 1. Time vs Power Usage
sns.lineplot(ax=axes[0], data=lego_rtx3050_8, x='Time_Minutes', y='Power_Watts', color='red', linewidth=2)
axes[0].set_title('Power Consumption Over Time')
axes[0].set_ylabel('Power (Watts)')

# 2. Time vs GPU Utilization
sns.lineplot(ax=axes[1], data=lego_rtx3050_8, x='Time_Minutes', y='GPU_Utilization_Pct', color='green', linewidth=2)
axes[1].set_title('GPU Utilization Over Time')
axes[1].set_ylabel('Utilization (%)')
axes[1].set_ylim(0, 105) # Cap at 100% (+5 for visual padding)

# 3. Time vs Memory Usage
sns.lineplot(ax=axes[2], data=lego_rtx3050_8, x='Time_Minutes', y='Memory_Used_MiB', color='blue', linewidth=2)
axes[2].set_title('VRAM Memory Usage Over Time')
axes[2].set_xlabel('Time (Minutes)')
axes[2].set_ylabel('Memory (MiB)')

plt.tight_layout()
plt.show()
```
